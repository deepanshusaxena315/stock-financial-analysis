```python
starting_price = data["Close"].iloc[0]
ending_price = data["Close"].iloc[-1]
total_return = ending_price/starting_price -1
total_return
```

```python
data["Cumulative Return"] = (1 + data["Daily Return"]).cumprod() - 1
data["Cumulative Return %"] = data["Cumulative Return"] * 100
data[["Close","Cumulative Return %"]].head()
```

```python
plt.figure(figsize = (12,6))
plt.plot(
    data["Cumulative Return %"],
    label = "AAPL Cumulative Return"
    )
plt.title("AAPL Cumulative Returns - Last 5 years")
plt.xlabel("Date")
plt.ylabel("Return (%)")
plt.grid()
plt.legend()
plt.show()
```

```python
data["MA50"] = data["Close"].rolling(window=50).mean()
data["MA200"] = data["Close"].rolling(window=200).mean()
```

```python
plt.figure(figsize=(14,7))
plt.plot(data["Close"],label= "AAPL Close")
plt.plot(data["MA50"],label= "50-Day Moving Average")
plt.plot(data["MA200"],label= "200-Day Moving Average")
plt.title("AAPL Price vs Moving Averages")
plt.xlabel("Days")
plt.ylabel("Price ($)")
plt.grid()
plt.legend()
plt.show()
```

```python
if isinstance(data.columns, pd.MultiIndex):
    data.columns = data.columns.get_level_values(0)
```

```python
data["Running Max"] = data["Close"].cummax()
data["DrawDown"] = ( data["Close"]/data["Running Max"]) - 1

max_drawdown = data["DrawDown"].min()
max_drawdown*100
```

```python
plt.figure(figsize=(14, 6))

plt.plot(data["DrawDown"] * 100)

plt.title("AAPL DrawDown")
plt.xlabel("Date")
plt.ylabel("DrawDown (%)")

plt.grid()

plt.show()
```

```python
best_day = data["Daily Return"].max()
worst_day = data["Daily Return"].min()
```

```python
summary = {
    "Total Return (%)": total_return * 100,
    "Average Daily Return (%)": average_daily_return * 100,
    "Best Day (%)": best_day * 100,
    "Worst Day (%)": worst_day * 100,
    "Annualized Volatility (%)": annualized_volatility * 100,
    "Maximum Drawdown (%)": max_drawdown * 100
}

summary
```

```python
summary_df = pd.DataFrame(
    summary.items(),
    columns=["Metric", "Value"]
)

summary_df
```

# 11. Multi-Stock Performance Comparison

In this section, we compare Apple (AAPL), Microsoft (MSFT), and NVIDIA (NVDA) using historical stock prices, cumulative returns, annualized volatility, maximum drawdown, and risk-adjusted performance.

All stocks are analyzed over a common historical period.

```python
tickers = ["AAPL","MSFT","NVDA"]
multi_data = yf.download(
    tickers,
    period="5y",
    auto_adjust=True,
    progress=False
)

multi_data.head()
```

```python

if isinstance(multi_data.columns, pd.MultiIndex):
    if "Close" in multi_data.columns.get_level_values(0):
        close_prices = multi_data["Close"]
    else:
        close_prices = multi_data.xs(
            "Close",
            axis=1,
            level=1
        )
else:
    close_prices = multi_data[["Close"]]

close_prices = close_prices.dropna()

close_prices.head()
```

```python
print(close_prices.shape)
print(close_prices.columns)
print(close_prices.isna().sum())
```

```python
daily_returns = close_prices.pct_change(fill_method=None)

daily_returns.head()
```

```python
initial_investment = 10_000

normalized_performance = (
    close_prices / close_prices.iloc[0]
)

investment_value = (
    normalized_performance * initial_investment
)

investment_value.tail()
```

```python
plt.figure(figsize=(14, 7))

for ticker in investment_value.columns:
    plt.plot(
        investment_value.index,
        investment_value[ticker],
        label=ticker
    )

plt.title("Growth of $10,000 — AAPL vs MSFT vs NVDA")
plt.xlabel("Date")
plt.ylabel("Investment Value ($)")

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

total_return = (
    close_prices.iloc[-1] / close_prices.iloc[0] - 1
) * 100

annualized_volatility = (
    daily_returns.std() * np.sqrt(252) * 100
)

running_max = close_prices.cummax()

drawdowns = (
    close_prices / running_max - 1
)

max_drawdown = drawdowns.min() * 100

sharpe_ratio = (
    daily_returns.mean()
    / daily_returns.std()
) * np.sqrt(252)

performance_summary = pd.DataFrame({
    "Total Return (%)": total_return,
    "Annualized Volatility (%)": annualized_volatility,
    "Maximum Drawdown (%)": max_drawdown,
    "Sharpe Ratio (0% risk-free)": sharpe_ratio
})

performance_summary.round(2)
```

```python

plt.figure(figsize=(14, 7))

for ticker in drawdowns.columns:
    plt.plot(
        drawdowns.index,
        drawdowns[ticker] * 100,
        label=ticker
    )

plt.title("Historical Drawdowns — AAPL vs MSFT vs NVDA")
plt.xlabel("Date")
plt.ylabel("Drawdown (%)")

plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 12. Correlation Analysis

Correlation measures how closely the daily returns of two stocks
move together. Values range from -1 to +1.

```python
correlation_matrix = daily_returns.corr()

correlation_matrix.round(3)
```

```python
plt.figure(figsize=(8, 6))

plt.imshow(
    correlation_matrix,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.yticks(
    range(len(correlation_matrix.index)),
    correlation_matrix.index
)

plt.title("Stock Return Correlation Matrix")

for i in range(len(correlation_matrix)):
    for j in range(len(correlation_matrix)):
        plt.text(
            j, i,
            f"{correlation_matrix.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.tight_layout()
plt.show()
```

```python
final_investment_value = investment_value.iloc[-1]

performance_summary["Ending Value ($)"] = (
    final_investment_value
)

performance_summary = performance_summary.sort_values(
    "Total Return (%)",
    ascending=False
)

performance_summary.round(2)
```
