# Stock market Analysis
Analysis of historical stock data using python.

**tools**: numpy,pandas,matplotlib,yfinance

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf
```

```python
ticker = "AAPL"
```

```python
data = yf.download(ticker,period = "5y")
```

```python
plt.figure(figsize = (12,6))
plt.plot(data["Close"],label="AAPL close")

plt.title("Apple Stock Price - Last 5 years")
plt.xlabel("Date")
plt.ylabel("Price ($)")

plt.legend()
plt.grid()
plt.show()
```

```python
data["Close"].head(10)
```

```python
data["Daily Return"] = data["Close"].pct_change()
```

```python
data.head()
```

```python
data["Daily Return %"] = data["Daily Return"] * 100
```

```python
data[["Close","Daily Return %"]].head(10)
```

```python
average_daily_return = data["Daily Return"].mean()
```

```python
average_daily_return*100
```

```python
print(data.loc[data["Daily Return"].idxmax()])
print("Worst Day:",data["Daily Return"].idxmin())
```

```python
plt.figure(figsize=(10,6))
plt.hist(data["Daily Return"].dropna(), bins=50)
plt.title("AAPL Daily Returns Distribution")
plt.xlabel("Daily Returns")
plt.ylabel("Frequency")
plt.show()
```

```python
daily_volatility = data["Daily Return"].std()
daily_volatility
```

```python
annualized_volatility = daily_volatility * np.sqrt(252)
annualized_volatility*100
```

```python
starting_price = data["Close"].iloc[0]
ending_price = data["Close"].iloc[-1]
total_return = ending_price/starting_price -1
total_return
```

```python
data["Cumulative Return"] = (1 + data["Daily Return"]).cumprod() - 1
data["Cumulative Return %"] = data["Cumulative Return"] * 100
data[["Close","Cumulative Return %"]].head()
```

```python
plt.figure(figsize = (12,6))
plt.plot(
    data["Cumulative Return %"],
    label = "AAPL Cumulative Return"
    )
plt.title("AAPL Cumulative Returns - Last 5 years")
plt.xlabel("Date")
plt.ylabel("Return (%)")
plt.grid()
plt.legend()
plt.show()
```

```python
data["MA50"] = data["Close"].rolling(window=50).mean()
data["MA200"] = data["Close"].rolling(window=200).mean()
```

```python
plt.figure(figsize=(14,7))
plt.plot(data["Close"],label= "AAPL Close")
plt.plot(data["MA50"],label= "50-Day Moving Average")
plt.plot(data["MA200"],label= "200-Day Moving Average")
plt.title("AAPL Price vs Moving Averages")
plt.xlabel("Days")
plt.ylabel("Price ($)")
plt.grid()
plt.legend()
plt.show()
```

```python
if isinstance(data.columns, pd.MultiIndex):
    data.columns = data.columns.get_level_values(0)
```

```python
data["Running Max"] = data["Close"].cummax()
data["DrawDown"] = ( data["Close"]/data["Running Max"]) - 1

max_drawdown = data["DrawDown"].min()
max_drawdown*100
```

```python
plt.figure(figsize=(14, 6))

plt.plot(data["DrawDown"] * 100)

plt.title("AAPL DrawDown")
plt.xlabel("Date")
plt.ylabel("DrawDown (%)")

plt.grid()

plt.show()
```

```python
best_day = data["Daily Return"].max()
worst_day = data["Daily Return"].min()
```

```python
summary = {
    "Total Return (%)": total_return * 100,
    "Average Daily Return (%)": average_daily_return * 100,
    "Best Day (%)": best_day * 100,
    "Worst Day (%)": worst_day * 100,
    "Annualized Volatility (%)": annualized_volatility * 100,
    "Maximum Drawdown (%)": max_drawdown * 100
}

summary
```

```python
summary_df = pd.DataFrame(
    summary.items(),
    columns=["Metric", "Value"]
)

summary_df
```
